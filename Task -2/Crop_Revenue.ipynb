```python
# Crop Yield and Revenue with Dictionaries

# Step 1 & 2: Create two dictionaries
yield_kg = {
    "Rice": 2500,
    "Wheat": 2000,
    "Maize": 3000,
    "Cotton": 1500
}

price_per_kg = {
    "Rice": 30,
    "Wheat": 25,
    "Maize": 20,
    "Cotton": 40
}

# Step 3: Create revenue dictionary using dictionary comprehension
revenue = {
    crop: yield_kg[crop] * price_per_kg[crop]
    for crop in yield_kg
}

# Step 4: Print every crop and its income
print("Crop Revenue:")

for crop, income in revenue.items():
    print(f"{crop}: ₹{income}")

# Step 5: Find the best crop
best_crop = max(revenue, key=revenue.get)

print("\nBest crop:", best_crop)
print("Highest income: ₹", revenue[best_crop])

# Step 6: Add a new crop
yield_kg["Sugarcane"] = 4000
price_per_kg["Sugarcane"] = 15

# Update one value
price_per_kg["Maize"] = 25

# Recalculate revenue
revenue = {
    crop: yield_kg[crop] * price_per_kg[crop]
    for crop in yield_kg
}

print("\nAfter changing the data:")

for crop, income in revenue.items():
    print(f"{crop}: ₹{income}")

# Safe lookup for a crop that does not exist
crop_name = "Barley"
price = price_per_kg.get(crop_name, "Crop not found")

print(f"\nSafe lookup for {crop_name}: {price}")
```

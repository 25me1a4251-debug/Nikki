```python
# Task 4: Advisory Mini-Program that Saves a Report

# Dictionary containing water requirements of crops
water_need_mm = {
    "Rice": 1200,
    "Wheat": 600,
    "Ragi": 500,
    "Maize": 700
}

# Function to give advice
def advise(crop, rainfall_mm):
    crop = crop.strip().title()

    if crop not in water_need_mm:
        return f"{crop}: No data available."

    gap = water_need_mm[crop] - rainfall_mm

    if gap > 0:
        return f"{crop}: Rainfall is {gap} mm below the water need. Additional water may be required."
    elif gap == 0:
        return f"{crop}: Rainfall exactly meets the water need. No additional water is required."
    else:
        return f"{crop}: Rainfall is {-gap} mm above the water need. Water supply is sufficient."


# List of crops to check
crops = ["Rice", " Ragi ", "Wheat", "Cotton"]

# Annual rainfall
annual_rainfall = 800

# Create and write the report file
with open("advisory_report.txt", "w") as file:

    for crop in crops:
        line = advise(crop, annual_rainfall)

        # Print on screen
        print(line)

        # Write to file
        file.write(line + "\n")


# Confirm the contents of the report
print("\n--- advisory_report.txt ---")

with open("advisory_report.txt", "r") as file:
    print(file.read())
```

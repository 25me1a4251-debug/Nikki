```python
# Task 3: Farm Records with a List of Dictionaries

# 1. Create the list of farm records
farms = [
    {"farmer": "Ravi", "crop": "ragi", "area_acre": 2, "yield_kg": 800},
    {"farmer": "Sita", "crop": " Paddy ", "area_acre": 4, "yield_kg": 1600},
    {"farmer": "Arun", "crop": "PADDY", "area_acre": 5, "yield_kg": 2000},
    {"farmer": "Meena", "crop": "ragi", "area_acre": 3, "yield_kg": None},
    {"farmer": "Kiran", "crop": "Paddy", "area_acre": 2, "yield_kg": -50},
    {"farmer": "Lakshmi", "crop": "ragi", "area_acre": 4, "yield_kg": 1200}
]

# 2. Create empty containers
by_crop = {}
bad_records = []

# 3. Process every farm
for farm in farms:

    # Clean crop name
    crop = farm["crop"].strip().title()

    # Validate yield
    if farm["yield_kg"] is None or farm["yield_kg"] < 0:
        bad_records.append(farm["farmer"])
        continue

    # Calculate yield per acre
    yield_per_acre = farm["yield_kg"] / farm["area_acre"]

    # Group yield by crop
    by_crop.setdefault(crop, [])
    by_crop[crop].append(yield_per_acre)

# 4. Print bad records
print("Bad records:", bad_records)

# 5. Print number of farms and average yield per acre
print("\nFarm statistics:")

for crop, yields in by_crop.items():
    number_of_farms = len(yields)
    average_yield = sum(yields) / number_of_farms

    print(f"{crop}:")
    print(f"  Number of farms = {number_of_farms}")
    print(f"  Average yield per acre = {average_yield:.2f} kg")
```
